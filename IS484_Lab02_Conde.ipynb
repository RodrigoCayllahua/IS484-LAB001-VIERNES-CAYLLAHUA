## Representación de un problema mediante estados

**Parte A: Completar la ficha del problema para el siguiente caso:** un repartidor debe ir desde el Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.

| Estado | Descripción |
| :--- | :--- |
| **Estado inicial** | El repartidor se encuentra en el Almacén |
| **Estado objetivo** | El repartidor ha llegado a la ZONA NORTE |
| **Acciones posibles** | El repartidor se puede desplazar a cualquier punto que esté conectado a él |
| **Costo de cada acción** | El costo de acción para los tramos va a ser el mismo |

---

**Parte B: Representar el mapa de la ciudad como un grafo en Python**

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}

print(mapa)
```

## Implementación de BFS

**Búsqueda en amplitud para encontrar el camino más corto en el mapa**

```python
def bfs(grafo, inicio, objetivo):
    # Inicializamos la cola con una lista que contiene el camino inicial: ['Almacen']
    cola = [[inicio]]
    
    # Creamos un conjunto (set) vacío para registrar los nodos visitados y evitar bucles
    visitados = set()
    
    # Bucle principal: se ejecuta mientras existan caminos pendientes en la cola
    while cola:
        # sacamos el primer camino ingresado (índice 0) para cumplir el orden FIFO (cola)
        camino = cola.pop(0)
        
        # Con [-1] tomamos el ÚLTIMO elemento de este camino (nuestra ubicación actual)
        nodo_actual = camino[-1]
        
        # Si la ubicación actual coincide con el destino final, retornamos el camino recorrido
        if nodo_actual == objetivo:
            return camino
            
        # Si el nodo actual no se ha visitado previamente, lo procesamos
        if nodo_actual not in visitados:
            visitados.add(nodo_actual)  # Marcamos el nodo como visitado
            
            # Recorremos cada nodo vecino conectado al nodo actual en el grafo
            for vecino in grafo[nodo_actual]:
                # Con el operador '+' unimos listas: creamos una nueva ruta extendida con el vecino
                nuevo_camino = camino + [vecino]
                
                # Agregamos la nueva ruta al final de la cola para evaluarla en los siguientes turnos
                cola.append(nuevo_camino)
                
    # Retornamos None si se revisaron todos los caminos posibles y no se halló ruta al objetivo
    return None

# Ejecutamos la función de búsqueda desde 'Almacen' hasta 'ZonaNorte'
resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (BFS):", resultado)
```

## Vectores y matrices con NumPy

**Parte A: Representar cuatro observaciones de un dataset simple** (por ejemplo, sensores de
temperatura y humedad) como un vector cada una, y juntarlas en una matriz:


```python
import numpy as np  # Importamos la librería NumPy para operaciones matriciales

# Definimos observaciones individuales como vectores 1D: [temperatura, humedad]
obs1 = np.array([22.5, 60])  # Registro 1
obs2 = np.array([19.0, 45])  # Registro 2
obs3 = np.array([25.3, 70])  # Registro 3
obs4 = np.array([21.0, 50])  # Registro 4

# Unimos los 4 vectores para estructurar una matriz 2D (Dataset)
dataset = np.array([obs1, obs2, obs3, obs4])

# Imprimimos la matriz completa y mostramos sus dimensiones (filas, columnas)
print("Dataset completo:")
print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)

# Seleccionamos la primera fila completa (índice 0) -> Observación 1
print("\nPrimera observación (fila 0):", dataset[0])

# Seleccionamos la primera columna completa con slicing [:, 0] -> Todas las temperaturas
print("Todas las temperaturas (columna 0):", dataset[:, 0])
```

## Producto escalar y distancia euclidiana

**Parte A:** Calcular el producto escalar entre dos observaciones


```python
# Calculamos el producto escalar entre la observación 1 y la observación 2
producto = np.dot(obs1, obs2)
print("Producto escalar entre obs1 y obs2:", producto)
```

**Parte B:** Calcular la distancia euclidiana entre dos observaciones, primero con la fórmula manual
y luego con NumPy

```python
# Calculamos la distancia euclidiana paso a paso (resta, cuadrado, suma y raíz)
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))
print("Distancia euclidiana (fórmula manual):", distancia_manual)

# Calculamos la misma distancia usando la función de norma vectorial de NumPy
distancia_numpy = np.linalg.norm(obs1 - obs2)
print("Distancia euclidiana (función NumPy):", distancia_numpy)
```
